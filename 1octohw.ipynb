### 1 wap to reverse given string

```python
s="Neha Mandilkar"
reverse=s[ : :-1]
print("The reverse of the string is:",reverse)
```

### 2 wap to check whether a string is palindrome

```python
s="535"
if s==s[::-1]:
    print("string is palindrome")
else:
    print("string is not palindrome")
```

### 3 Write a program to count how many times each vowel occurs in a string.

```python
s ="neha mandilkar"
a=0
e=0
i=0
o=0
u=0

for ch in s:
    if ch in "a":
        a+=1
    elif ch in "e":
        e+=1
    elif ch in "i":
        i+=1
    elif ch in "o":
        o+=1
    elif ch in "u":
        u+=1
print("a=",a)
print("e=",e)
print("i=",i)
print("o=",o)
print("u=",u)
```

### 6 wap to check  count of the blank spaces

```python
s="python is a programming language"
count=0
for ch in s:
    if ch==' ':
        count+=1
print("count of the blank spaces",count)
```
